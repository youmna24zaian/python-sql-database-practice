#### **Exercise 1**
**Retrieve the following information for every order:** 
- FirstName 
- LastName 
- OrderID 
- OrderDate 
- Amount 
- Status 

Display only orders whose amount is greater than 800.


```python
cursor.execute("""
SELECT C.FirstName,
C.LastName,
O.OrderID,
O.OrderDate,
O.Amount,
O.Status
FROM Customers C
JOIN Orders O
    ON C.CustomerID = O.CustomerID

WHERE Amount > 800
""")

cursor.fetchall()
```

#### **Exercise 2**
Retrieve the CustomerID, FirstName, and LastName of customers who have never placed an order. 

Solve it using LEFT JOIN.

```python
cursor.execute("""
SELECT C.CustomerID,
C.FirstName,
C.LastName,
O.OrderID
FROM Customers C
LEFT JOIN Orders O
    ON C.CustomerID = O.CustomerID
WHERE O.OrderID IS NULL
""")

cursor.fetchall()
```

#### **Exercise 3**
**Create a View called CitySalesSummary that shows, for each city:** 

- City 
- Number of customers who placed orders 
- Number of orders 
- Total order amount 
- Average order amount 
- Highest order amount 

**Then query the View to retrieve only cities that:** 

- have at least 2 orders, and have a total order amount greater than the average total order amount across all cities. 

```python
cursor.execute("""
CREATE OR ALTER VIEW dbo.CitySalesSummary
AS
SELECT C.City,
COUNT(DISTINCT C.CustomerID) AS 'Number of customers',
COUNT(O.OrderID) AS 'Number of orders',
SUM(O.Amount) AS 'Total order amount',
AVG(O.Amount) AS 'Average order amount',
MAX(O.Amount) AS 'Highest order amount'
FROM Customers C
JOIN Orders O
    ON C.CustomerID = O.CustomerID
GROUP BY C.City
""")
```

```python
cursor.execute("""
SELECT City
FROM dbo.CitySalesSummary
WHERE [Number of orders] >= 2 AND [Total order amount] > (SELECT AVG([Total order amount]) FROM dbo.CitySalesSummary)
""")

cursor.fetchall()
```

#### **Exercise 4**
**Create a CTE called CustomerSpending that calculates, for each customer:** 

- CustomerID 
- FirstName 
- LastName 
- Number of orders 
- Total amount spent 
- Average order amount 

Then retrieve the top 3 customers whose total spending is greater than the average total spending of all customers who placed orders.

Order the result from highest to lowest TotalSpent.


```python
cursor.execute("""
WITH CustomerSpending AS
(
SELECT C.CustomerID,
C.FirstName,
C.LastName,
COUNT(O.OrderID) AS 'Number of orders',
SUM(O.Amount) AS 'Total amount spent',
AVG(O.Amount) AS 'Average order amount'
FROM Customers C
JOIN Orders O
    ON C.CustomerID = O.CustomerID
GROUP BY C.CustomerID,
C.FirstName,
C.LastName
)
SELECT TOP 3 CustomerID, FirstName, LastName
FROM CustomerSpending
WHERE [Total amount spent] > (SELECT AVG([Total amount spent]) FROM CustomerSpending)
ORDER BY [Total amount spent] DESC
""")

cursor.fetchall()
```

## **Exercise 5**
**Create a CTE that calculates, for each customer who has placed orders:**

- CustomerID 
- FirstName 
- LastName 
- NumberOfOrders 
- TotalSpent 
- AverageOrderAmount 
- MaximumOrderAmount 

**Then retrieve only customers who satisfy all of the following:**
- placed at least 2 orders, 
- have a total spending greater than the average total spending of all customers who placed orders, and have at least one order whose amount is greater than the overall average order amount in the Orders table.

```python
cursor.execute("""
WITH CustomerOrderStats AS
(
SELECT C.CustomerID,
C.FirstName,
C.LastName,
COUNT(O.OrderID) AS 'NumberOfOrders',
SUM(O.Amount) AS 'TotalSpent',
AVG(O.Amount) AS 'AverageOrderAmount',
MAX(O.Amount) AS 'MaximumOrderAmount'
FROM Customers C
JOIN Orders O
    ON C.CustomerID = O.CustomerID
GROUP BY C.CustomerID,
C.FirstName,
C.LastName
)
SELECT CustomerID,
FirstName,
LastName
FROM CustomerOrderStats
WHERE NumberOfOrders >= 2 AND TotalSpent > (SELECT AVG(TotalSpent) 
FROM CustomerOrderStats) AND 
MaximumOrderAmount > (SELECT AVG(O.Amount)FROM Orders O)
""")
```
