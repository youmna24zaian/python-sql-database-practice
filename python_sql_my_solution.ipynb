## **Topic 5 — Connecting Python to SQL Server**

````python
import pyodbc
````

### **Connecting Python to SQL Server**

#### **1. `pyodbc.connect()`**

```python
connection = pyodbc.connect(...)
```

`pyodbc.connect()` opens a connection between Python and a database.

The `connection` variable stores the database connection.

---

#### **2. DRIVER**

```python
'DRIVER={ODBC Driver 18 for SQL Server};'
```

Specifies the ODBC driver that Python should use to communicate with SQL Server.

**ODBC** = Open Database Connectivity.

The driver acts as a bridge between `pyodbc` and SQL Server.

---

#### **3. SERVER**

```python
f'SERVER={server};'
```

Specifies the SQL Server instance that Python should connect to.

**Example:**

```python
server = r'DESKTOP-ABC123\SQLEXPRESS'
```

The `f` before the string allows the value of the `server` variable to be inserted into the string.

---

#### **4. DATABASE**

```python
f'DATABASE={database};'
```

Specifies which database on the SQL Server should be used.

**Example:**

```python
database = 'CompanyDB'
```

This tells Python to connect specifically to the `CompanyDB` database.

---

#### **5. Trusted_Connection**

```python
'Trusted_Connection=yes;'
```

Tells the driver to use **Windows Authentication**.

The current Windows user account is used instead of providing a SQL Server username and password in the connection string.

---

#### **6. Encrypt**

```python
'Encrypt=yes;'
```

Requests an encrypted connection between Python and SQL Server.

This helps protect data while it is being transmitted.

---

#### **7. TrustServerCertificate**

```python
'TrustServerCertificate=yes;'
```

Tells the driver to trust the SQL Server's certificate without requiring it to be verified by a trusted certificate authority.

This is commonly useful in local/development environments where SQL Server may use a self-signed certificate.

---

#### **8. Semicolon `;`**

The semicolon separates different connection settings.

**For example:**

```text
DRIVER=...;
SERVER=...;
DATABASE=...;
Encrypt=yes;
```

Each setting is separated by `;`.

---

### **Complete Example Complete Example**

```text
Python
   ↓
pyodbc
   ↓
ODBC Driver 18
   ↓
SQL Server
   ↓
CompanyDB
```

After the connection is established, a cursor can be created:

```python
cursor = connection.cursor()
```

The cursor is then used to execute SQL queries.

````python
server = r'YOUR_SERVER\SQLEXPRESS'
database = 'Online_Course'

connection = pyodbc.connect(
    'DRIVER={ODBC Driver 18 for SQL Server};'
    f'SERVER={server};'
    f'DATABASE={database};'
    'Trusted_Connection=yes;'
    'Encrypt=yes;'
    'TrustServerCertificate=yes;'
)

cursor = connection.cursor()
print("Connected Successfully!")
````

## **Topic 8 — Joining student and enrollment records**

## **Topic 10 — Conditional labels with SQL CASE**

#### **Query each student's course and grade, and classify the grade as:**
- Excellent for grades of at least 90.
- Very Good for grades of at least 80.
- Good for grades of at least 70.
- Pass for grades of at least 50.
- Fail otherwise.

````python
query = """
SELECT StudentID, CourseName, Grade,
    CASE 
        WHEN Grade >= 90 THEN 'Excellent'
        WHEN Grade >= 80 THEN 'Very Good'
        WHEN Grade >= 70 THEN 'Good'
        WHEN Grade >= 50 THEN  'Pass'
        ELSE 'Fail'
    END AS GPA
FROM Enrollments
"""

pd.read_sql(query, connection)
````

## **Topic 11 — Joining data and classifying pass or fail**

````python
query = """ 
SELECT s.StudentID, 
s.FirstName,
s.LastName,
e.CourseName,
e.Grade,
CASE
    WHEN e.Grade >= 50 THEN 'Passed'
    ELSE 'Failed'
END AS Result
FROM Students s
JOIN Enrollments e
    ON s.StudentID = e.StudentID
"""
pd.read_sql(query, connection)
````
